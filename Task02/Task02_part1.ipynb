```python
from decimal import Decimal

# 1,000 payments of Rs 33.40 and three small adjustments of Rs 0.10
payments = ["33.40"] * 1000 + ["0.10"] * 3

def total_float(items):
    total = 0.0
    for amount in items:
        total += float(amount)
    return total

def total_decimal(items):
    total = Decimal("0")
    for amount in items:
        total += Decimal(amount)
    return total

print("float :", total_float(payments))
print("Decimal:", total_decimal(payments))
print(0.1 + 0.2 == 0.3, Decimal("0.1") + Decimal("0.2") == Decimal("0.3"))
```

```python
import sys
from decimal import Decimal

payments = ["33.40"] * 1000 + ["0.10"] * 3

# Method 1: integer paise
def total_paise(items):
    total = 0
    for amount in items:
        total += round(float(amount) * 100)
    return total

paise = total_paise(payments)
print("Integer paise:", paise)
print("In rupees    :", paise / 100)

# Method 2: float with round() after every addition
def total_float_round(items):
    total = 0.0
    for amount in items:
        total = round(total + float(amount), 2)
    return total

print("Float + round:", total_float_round(payments))

# Memory per value
print("int     :", sys.getsizeof(3340), "bytes")
print("float   :", sys.getsizeof(33.40), "bytes")
print("Decimal :", sys.getsizeof(Decimal("33.40")), "bytes")
```
